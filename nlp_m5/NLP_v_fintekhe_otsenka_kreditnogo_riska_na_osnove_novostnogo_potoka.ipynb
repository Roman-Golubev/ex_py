## 1. Что такое финтех

**Финтех** — это применение технологических инноваций к финансовым услугам, продуктам и процессам.

Совет по финансовой стабильности, Financial Stability Board, определяет финтех через технологические инновации, которые могут менять финансовые бизнес-модели, приложения, процессы и продукты. Поэтому финтех — не только мобильные приложения и стартапы. Технологическое решение внутри банка, страховой компании или инвестиционного фонда также может относиться к этой области. [Источник: FSB](https://www.fsb.org/2019/02/fsb-report-assesses-fintech-developments-and-potential-financial-stability-implications/).

## 3. Дефолт и вероятность дефолта

### Что считать дефолтом

**Дефолт** — событие неисполнения обязательств, определённое правилами конкретной задачи или системы оценки.

В Базельском подходе к основаниям признания дефолта относятся, в частности, маловероятность полного исполнения обязательств без обращения к обеспечению либо просрочка более 90 дней по существенному обязательству. Для отдельных требований предусмотрены особенности. Таким образом, дефолт не сводится только к банкротству или только к достижению определённого срока просрочки. [Basel Framework, CRE36](https://www.bis.org/committees/bcbs/basel-framework/standard/cre/36/inforce/2023-01-01/published/2022-12-08).

Для исследования определение целевого события необходимо зафиксировать до обучения модели. Нужно решить, какие обязательства учитываются, какая дата считается датой дефолта, как обрабатываются технические нарушения, реструктуризации и повторные события.

**Банкротство, просрочка, реструктуризация и дефолт — связанные, но не тождественные понятия.** Модель, обученная на одном из этих событий, не становится автоматически моделью любого другого.

### Что такое PD

**Вероятность дефолта (PD)** — вероятность того, что компания допустит дефолт в течение заданного периода, с учётом информации, доступной на дату оценки.

$$
\mathrm{PD}_{i,t}^{(H)} = \mathrm{P}\left(Y_{i,t}^{(H)} = 1 \mid I_{i,t}\right)
$$

Здесь $i$ — компания, $t$ — дата оценки, $H$ — горизонт прогнозирования.

Индикатор $Y_{i,t}^{(H)}$ равен 1, если компания допускает дефолт в течение этого горизонта, и 0 — в противном случае.

$I_{i,t}$ — информация о компании, доступная на дату оценки. Символ $\mid$ читается как «при условии».

Например, годовая PD, равная 3%, означает оценённую вероятность дефолта компании в течение следующего года. Это не означает, что компания вернёт только 97% долга или что потери обязательно составят 3% суммы кредита.

В формулировке задачи должны быть определены **объект оценки, событие и горизонт**. Без них число, названное PD, невозможно однозначно интерпретировать.

### Пример: ожидаемые кредитные потери

Для иллюстрации используем упрощённую модель ожидаемых потерь:

$$
\mathrm{EL}=\mathrm{PD}\times\mathrm{LGD}\times\mathrm{EAD}.
$$

Здесь **EL**, expected loss, — ожидаемые потери; **LGD**, loss given default, — доля потерь при дефолте; **EAD**, exposure at default, — величина требования, подверженного риску на момент дефолта.

В этом учебном расчёте EAD фиксирована, а LGD задана как ожидаемая доля потерь при дефолте. Формула не учитывает все особенности реального расчёта: изменение задолженности, время дефолта и возвратов, сценарии и дисконтирование.

Предположим, что EAD составляет 100 млн рублей, годовая PD — 2%, а LGD — 40%. Тогда:

$$
\mathrm{EL}=100\,000\,000\times0.02\times0.40=800\,000.
$$

Если при остальных неизменных предположениях PD составляет 10%, получим:

$$
\mathrm{EL}=100\,000\,000\times0.10\times0.40=4\,000\,000.
$$

Разница в оценке ожидаемых потерь — 3,2 млн рублей. Поэтому качество оценки риска влияет на представление об экономике сделки.

## 5. Регуляторные подходы и назначение модели

### Стандартизированный подход, F-IRB и A-IRB

Для новостной модели объяснение может состоять из последовательности:

**Исходная публикация → выделенное событие → признак компании → влияние на оценку.**

Например, аналитик должен иметь возможность увидеть, какие сообщения привели к сигналу об ухудшении положения заёмщика и действительно ли они относятся к нужному юридическому лицу.

Интерпретируемость не равна простоте формулы. Сложная система может предоставлять проверяемые промежуточные результаты, а простая — использовать ошибочно сформированные признаки. Вместе с тем правдоподобное объяснение не заменяет проверки прогнозного качества.

## 6. Кредитование бизнеса и классический скоринг

Далее мы сосредоточимся на компаниях. Для корпоративного кредитования недостаточно знать, что предприятие прибыльно: необходимо оценить его способность выполнять обязательства с учётом денежных потоков, долговой нагрузки, сроков платежей и устойчивости бизнеса.

Представим два возможных решения банка. При недооценке риска он может выдать кредит на условиях, не соответствующих вероятным потерям. При переоценке — отказаться от приемлемого заёмщика. Следовательно, качественная оценка нужна не только для ограничения потерь, но и для обоснованного предоставления финансирования.

### Как возник скоринг

До широкого распространения статистических моделей кредитные решения в значительной степени опирались на индивидуальный анализ заявок и экспертное суждение. Один из ранних этапов развития формализованной оценки — исследование **David Durand 1941 года**, посвящённое факторам риска потребительского кредитования и формулам оценки заёмщиков. Это историческая веха, а не единственная дата изобретения скоринга. [Durand, Risk Elements in Consumer Instalment Financing](https://www.nber.org/books-and-chapters/risk-elements-consumer-instalment-financing).

В корпоративном анализе известна работа **Edward Altman 1968 года**, использующая финансовые коэффициенты и дискриминантный анализ для прогнозирования банкротства. С ней связан Z-score. Такой балл не является автоматически калиброванной годовой PD для любой страны, отрасли и периода. [Altman, Financial Ratios, Discriminant Analysis and the Prediction of Corporate Bankruptcy](https://doi.org/10.1111/j.1540-6261.1968.tb00843.x).

Развитие моделей не отменило экспертное рассмотрение. Модель и аналитик могут выполнять разные части одной задачи.

### Скоринг, рейтинг и PD: в чём различие

**Скоринг** — преобразование информации об объекте в количественную оценку по определённым правилам или модели.

**Кредитный рейтинг** — категория кредитного качества заёмщика или обязательства. Рейтинги могут присваиваться внешним агентством либо использоваться во внутренней системе банка.

**PD** — вероятность конкретного дефолта на заданном горизонте.

Балльная оценка может использоваться для присвоения рейтинга, а рейтинговый класс — сопоставляться с PD. Но эти переходы требуют правил и калибровки. Порядок компаний по риску и точность оценки вероятностей — разные характеристики модели. Требования к рейтинговым системам и оценкам компонентов риска рассматриваются отдельно и в [Basel Framework](https://www.bis.org/committees/bcbs/basel-framework/standard/cre/36/inforce/2023-01-01/published/2022-12-08).

## 7. Альтернативный скоринг и источники данных

В рамках практикума будем называть **альтернативным скорингом** оценку кредитного риска с использованием дополнительных источников по отношению к выбранной традиционной базе.

Альтернативность определяется прежде всего данными, а не сложностью алгоритма. Нейросеть на стандартных финансовых коэффициентах не становится альтернативной только из-за архитектуры. И наоборот, простая модель может использовать нетрадиционный источник информации.

Среди примеров — сведения о торговле, электронной коммерции, цепочках поставок и платежах. Они могут использоваться при оценке малого и среднего бизнеса. CDI обеспечивает обмен данными, но не задаёт единую модель кредитного решения. [Описание CDI](https://cdi.hkma.gov.hk/), [официальное сообщение о запуске](https://www.info.gov.hk/gia/general/202210/24/P2022102400237.htm).

**Индия: Account Aggregator и GSTN.** В инфраструктуру обмена финансовой информацией включены данные Goods and Services Tax Network. Один из заявленных сценариев — кредитование малого и среднего бизнеса на основе денежных потоков. Здесь существенны не только алгоритмы, но и правила доступа, согласия и передачи информации. [Описание GSTN](https://gstn.org.in/account-aggregator).

## 8. От анализа новостей к автоматизированной оценке

Представим компанию с приемлемыми показателями по последней доступной отчётности. Затем появляются сообщения о потере крупного заказчика, остановке производства и переговорах с кредиторами. Это учебный пример: сами по себе сообщения не доказывают будущего дефолта, но дают основания проверить устойчивость денежных потоков и способность обслуживать долг.

В процессе, предусматривающем экспертные корректировки, аналитик может использовать подобную информацию для дополнительной проверки или пересмотра оценки. Автоматизированная система выполняет часть этой работы: находит сообщения, определяет компанию, классифицирует события и формирует показатель риска.



## 9. Что показывают исследования текстов и кредитного риска

При чтении литературы важно различать источники текстов и целевые переменные. Прогноз наблюдавшегося дефолта, объяснение кредитного спреда, прогноз банкротства и воспроизведение сторонней модельной PD — разные задачи.

### 9.1. Темы новостей и дефолты облигаций

**Tang, W., Bu, H., Zuo, Y., Wu, J. (2024). [Unlocking the power of the topic content in news headlines: BERTopic for predicting Chinese corporate bond defaults](https://doi.org/10.1016/j.frl.2024.105062). Finance Research Letters, 62, 105062.**

Авторы анализируют новостные заголовки за 2014–2021 годы и используют BERTopic для получения тематических признаков. Из 78 исходных тем формируются 18 экономически интерпретируемых показателей. Они добавляются к финансовым коэффициентам и экономическим переменным в моделях первого дефолта корпоративных облигаций Китая на трёхмесячном горизонте.

В исследовании тематическое содержание новостей даёт дополнительную прогностическую информацию. Для нашего практикума важна конструкция «темы текстов → показатели компании → прогноз события».

Результат не означает, что те же темы или модель дадут аналогичный эффект на российском рынке и другом горизонте. Это гипотеза, которую требуется проверить на собственных данных.

### 9.2. Новости и рыночная оценка кредитного риска

**Tsai, F.-T., Lu, H.-M., Hung, M.-W. (2016). [The impact of news articles and corporate disclosure on credit risk valuation](https://doi.org/10.1016/j.jbankfin.2016.03.018). Journal of Banking & Finance, 68, 100–116.**

Авторы исследуют связь новостного покрытия, тональности и корпоративных раскрытий с оценкой риска на рынке кредитных дефолтных свопов, CDS. CDS — инструмент передачи кредитного риска; его цена отражает рыночную оценку соответствующей защиты.

На 13-летних данных текстовые характеристики содержат дополнительную информацию для оценки кредитного риска. Однако объяснение цены CDS не тождественно прогнозированию фактического дефолта: рыночные котировки отражают не только вероятность события.



### 9.3. Текстовые раскрытия и банкротства

**Mai, F., Tian, S., Lee, C., Ma, L. (2019). [Deep learning models for bankruptcy prediction using textual disclosures](https://doi.org/10.1016/j.ejor.2018.10.024). European Journal of Operational Research, 274(2), 743–758.**

Авторы используют данные 11 827 американских публичных компаний и исследуют прогноз банкротства по текстовым раскрытиям. В их эксперименте объединение текста с финансовыми и рыночными переменными улучшает прогнозирование. [Описание работы в университете авторов](https://researchwith.stevens.edu/en/publications/deep-learning-models-for-bankruptcy-prediction-using-textual-disc/).

Для нас важна идея взаимодополняемости разных типов данных. Текст не обязательно должен заменить финансовые показатели: он может добавлять информацию к ним.

При этом источник здесь — раскрытия компаний, а не журналистские новости. Целевое событие — банкротство. Эти различия необходимо учитывать при переносе постановки.

### 9.4. Открытая мультимодальная постановка

**Arno, H., Mulier, K., Baeck, J., Demeester, T. [From Numbers to Words: Multi-Modal Bankruptcy Prediction Using the ECL Dataset](https://arxiv.org/abs/2401.12652). FinNLP 2023; версия на arXiv — 2024.**

Работа предлагает набор данных с числовой и текстовой информацией из 10-K и метками банкротства. Авторы обнаруживают взаимодополняемость модальностей, но отмечают ограничение: модель может распознавать финансовую слабость, не умея хорошо различать банкротство в следующем году и в более поздние годы.

В рассмотренных экспериментах прямое zero-shot-прогнозирование банкротства через GPT давало слабый результат, хотя модель могла извлекать содержательные резюме.

Это подводит к важному вопросу: **мы прогнозируем событие на заданном горизонте или просто узнаём неблагополучную компанию?**

### 9.5. Разногласие между информационными источниками

**Garcia, J. (2025). [Beyond the headlines: Sentiment divergence and financial distress](https://doi.org/10.1016/j.gfj.2025.101126). Global Finance Journal, 66, 101126.**

На данных 1 823 американских компаний исследуется расхождение между тональностью традиционных новостей и Twitter, а также изменчивость этого расхождения. Автор обнаруживает, что эти характеристики связаны с финансовым риском по-разному.

Для разработки признаков полезна идея учитывать не только среднюю тональность, но и согласованность источников. Однако основная зависимая переменная в работе — годовая модельная PD Bloomberg, а не непосредственно наблюдавшийся дефолт. Поэтому результаты нельзя представлять как независимую проверку прогноза реальных дефолтов.

## 12. Новости в других финансовых задачах

Обработка текстов, привязка к объектам и датам, построение признаков и их агрегация применимы не только к кредитному риску. Меняются экономический вопрос, целевая переменная и требования к проверке результата.

### 12.1. M&A: слияния и поглощения

В исследованиях **M&A — mergers and acquisitions** можно изучать вероятность объявления сделки, её условия, завершение или срыв, а также реакцию рынка. Эти постановки требуют разных выборок и точек отсчёта.

**Ahern, K. R., Sosyura, D. (2014). [Who Writes the News? Corporate Press Releases during Merger Negotiations](https://doi.org/10.1111/jofi.12109). Journal of Finance, 69(1), 241–291.**

Авторы исследуют корпоративные коммуникации во время переговоров о слияниях с оплатой акциями. Компании-покупатели увеличивают выпуск собственных новостей после начала непубличных переговоров, но до объявления сделки. Такая активность связана с временным ростом цены акций в период определения условий обмена.

Важный вывод: новостной поток не всегда является независимым наблюдением. Его участники могут стратегически влиять на содержание и время публикаций. Это исследование коммуникационной стратегии, а не готовая модель предсказания M&A.

**Liao, R., Wang, X., Wu, G. (2021). [The Role of Media in Mergers and Acquisitions](https://doi.org/10.1016/j.intfin.2021.101299). Journal of International Financial Markets, Institutions and Money, 74, 101299.**

Авторы анализируют 77 552 завершённые сделки за 2000–2015 годы и исследуют связь медийной тональности с участием в приобретениях, премиями и доходностями. [Описание работы в университете авторов](https://scholarship.richmond.edu/finance-faculty-publications/63/).

Возможное продолжение нашего практикума — прогноз завершения уже объявленной сделки. Однако для этого нужны не только завершённые, но и сорванные или отозванные сделки. Выборка только успешных сделок не позволяет обучить такой классификатор.

### 12.2. Реакция рынка: содержание и внимание

**Tetlock, P. C. (2007). [Giving Content to Investor Sentiment: The Role of Media in the Stock Market](https://doi.org/10.1111/j.1540-6261.2007.01232.x). Journal of Finance, 62(3), 1139–1168.**

На материале ежедневной колонки Wall Street Journal автор изучает связь пессимизма в СМИ с рынком. Высокий пессимизм связан с краткосрочным давлением на цены и последующим разворотом, а крайние значения тональности — с повышенным объёмом торгов.

Для текстовой модели это важное различие: признак может отражать настроение и внимание участников, а не только изменения финансового состояния компании.

**Fedyk, A. (2024). [Front-Page News: The Effect of News Positioning on Financial Markets](https://doi.org/10.1111/jofi.13287). Journal of Finance, 79(1), 5–33.**

Исследование использует вариацию заметности публикаций в Bloomberg. Для новостей на первой странице объём торгов в первые десять минут был на 240% выше, чем для сопоставимых по значимости менее заметных публикаций. Результат относится к конкретному исследовательскому дизайну.

Таким образом, важны не только слова, но и распространение информации. В проекте о рыночной реакции потребуется учитывать точное время публикации, торговые часы и скорость включения информации в цены. Для вывода об инвестиционной применимости также нужны расчёты после транзакционных издержек.

### 12.3. Неопределённость и волатильность

**Manela, A., Moreira, A. (2017). [News implied volatility and disaster concerns](https://doi.org/10.1016/j.jfineco.2016.01.032). Journal of Financial Economics, 123(1), 137–162.**

Авторы строят текстовый показатель неопределённости **NVIX — news implied volatility** по материалам первой страницы Wall Street Journal. Полученный исторический ряд начинается в 1890 году. Исследование связывает новостную неопределённость с опасениями редких неблагоприятных событий и финансовыми рынками.

Методологическая идея состоит в том, что тексты позволяют измерить экономическую характеристику на периодах, для которых привычного рыночного показателя ещё не существовало.

Возможный проект — новостной индекс отраслевой неопределённости и проверка его связи с будущей волатильностью. Это уже задача построения временного показателя, а не классификации компаний по дефолту.

### 12.4. Деловые циклы и экономическая активность

**Bybee, L., Kelly, B., Manela, A., Xiu, D. (2024). [Business News and Business Cycles](https://doi.org/10.1111/jofi.13377). Journal of Finance, 79(5), 3105–3147.**

Авторы используют около 800 тысяч статей Wall Street Journal за 1984–2017 годы. Тематическая модель превращает новости в интерпретируемые показатели распределения внимания между экономическими темами. Эти показатели помогают анализировать и прогнозировать экономическую динамику и финансовые рынки.

Здесь тема текста становится экономической переменной. Например, можно исследовать, содержит ли изменение интенсивности обсуждения спроса, инвестиций или логистики дополнительную информацию о будущей активности отрасли.

При проверке такого проекта необходимо учитывать даты публикации статистики и её последующие пересмотры. Иначе в исторический эксперимент может попасть информация, которой на дату прогноза ещё не было.

### 12.5. Неопределённость экономической политики

**Baker, S. R., Bloom, N., Davis, S. J. (2016). [Measuring Economic Policy Uncertainty](https://doi.org/10.1093/qje/qjw024). Quarterly Journal of Economics, 131(4), 1593–1636.**

Авторы предлагают новостной индекс **EPU — economic policy uncertainty** и исследуют его связь с финансовыми и макроэкономическими показателями. Конструкция индекса сопровождается проверками, в том числе на вручную изученных статьях.

Работа показывает, что исследовательский вклад не обязательно требует самой сложной модели. Существенны ясное определение измеряемого понятия и проверка того, что текстовый показатель действительно ему соответствует.

EPU измеряет обсуждаемую неопределённость, а не качество экономической политики и не вероятность дефолта государства.

### 12.6. Геополитический риск

**Caldara, D., Iacoviello, M. (2022). [Measuring Geopolitical Risk](https://doi.org/10.1257/aer.20191823). American Economic Review, 112(4), 1194–1225.**

Авторы создают новостной индекс **GPR — geopolitical risk**, различая обсуждение угроз и реализацию неблагоприятных событий. В исследовании анализируется связь индекса с инвестициями, занятостью и экономической динамикой. [Страница статьи у издателя](https://www.aeaweb.org/articles?id=10.1257/aer.20191823).

Для NLP-задачи принципиален переход от абстрактного понятия «риск» к правилам измерения: какие сообщения считаются релевантными, как различаются типы событий и как проверяется экономическая содержательность результата.

Для сравнения стран или регионов необходимо учитывать различия в языках, источниках и интенсивности освещения. Большее число публикаций само по себе не доказывает больший объективный риск.

### 12.7. Климатические новости и хеджирование

**Engle, R. F., Giglio, S., Kelly, B., Lee, H., Stroebel, J. (2020). [Hedging Climate Change News](https://doi.org/10.1093/rfs/hhz072). Review of Financial Studies, 33(3), 1184–1216.**

Авторы строят новостной климатический показатель и портфели, предназначенные для хеджирования его изменений. Тексты используются для определения риска, с которым затем связывается финансовый инструмент. [Рабочая версия NBER](https://www.nber.org/papers/w25734).

Это пример того, как NLP становится частью задачи управления портфелем. При этом хеджирование изменений новостного индекса не означает устранение физических климатических потерь или всех связанных с климатом рисков.

## 13. Постановка нашего практикума

В нашем кейсе рассматриваются компании, идентифицированные по ИНН, связанные с ними новости и информация о последующем дефолте. Цель — получить из новостей признаки и проверить их полезность для прогнозирования, построить PD модель.

Общая схема:

**Новости о компании → обработка текста → признаки новостей → агрегация по компании и времени → модель кредитного риска.**

Объект итогового прогноза — **компания на дату оценки**, а не отдельная публикация.

Целевая переменная равна единице, если заданный дефолт произошёл на прогнозном горизонте, и нулю — если компания наблюдалась до конца этого горизонта без такого события.

Все признаки должны строиться только по информации, доступной на дату оценки. Сообщение, опубликованное после неё, нельзя использовать для исторического прогноза, даже если оно рассказывает о более ранних обстоятельствах.


По общей логике мы создаём учебное аналитическое ядро, близкое к рассмотренным коммерческим решениям. Однако сбор и лицензирование источников, непрерывное обновление, контроль качества и интеграция в банковские процессы остаются самостоятельными задачами промышленного внедрения.

## 17. Анализ тональности на основе частот слов

Анализ тональности (*sentiment analysis*) — задача определения положительной, отрицательной или нейтральной оценки, выраженной в тексте. В этом подходе рассматривается **бинарная классификация**: положительная и отрицательная тональность.

Основная идея состоит в том, что слова встречаются в текстах разных классов с разной частотой. Например, слова «успех» и «улучшение» могут чаще встречаться в положительных сообщениях, а «проблема» и «ухудшение» — в отрицательных. Эти закономерности извлекаются из размеченных обучающих данных.

Каждый заголовок представляется **вектором длины три**: константой и двумя признаками, построенными на частотах его слов в положительных и отрицательных текстах.

### 1. Подготовка текста

Перед подсчётом частот применяется одинаковая предобработка обучающих и новых текстов:

- приведение к нижнему регистру;
- разбиение на токены;
- при необходимости — лемматизация или стемминг.

Например, после лемматизации заголовок:

> «Прибыль компании выросла»

может быть представлен последовательностью:

```text
прибыль, компания, вырасти
```

Удалять стоп-слова следует осмотрительно: частица «не» может менять смысл высказывания и важна для тональности.

### 2. Построение словаря частот

Обучающая выборка состоит из текстов с известными метками:

$$
y_i=
\begin{cases}
1, & \text{положительная тональность},\\
0, & \text{отрицательная тональность}.
\end{cases}
$$

Для каждого слова $w$ подсчитывается число его вхождений в тексты каждого класса:

$$
\operatorname{freq}(w,c)=
\sum_{i:\,y_i=c}\operatorname{count}(w,h_i),
$$

где:

- $h_i$ — обучающий текст;
- $\operatorname{count}(w,h_i)$ — число вхождений слова $w$ в этот текст;
- $c$ — класс тональности.

Словарь хранит соответствие:

$$
(\text{слово},\text{класс})\rightarrow\text{частота}.
$$

**Словарь строится только на обучающей выборке.** Тестовые тексты и их метки при подсчёте частот не используются.

### 3. Представление заголовка вектором

Пусть после предобработки заголовок состоит из токенов:

$$
h=(w_1,w_2,\ldots,w_m).
$$

Его представление имеет вид:

$$
\mathbf{x}(h)=
\left[
1,\;
\sum_{j=1}^{m}\operatorname{freq}(w_j,1),\;
\sum_{j=1}^{m}\operatorname{freq}(w_j,0)
\right].
$$

Координаты вектора:

- **$x_0=1$** — константа для свободного члена модели;
- **$x_+$** — сумма частот слов заголовка в положительных обучающих текстах;
- **$x_-$** — сумма частот этих слов в отрицательных обучающих текстах.

Если слово отсутствует в словаре, его частота принимается равной нулю. Повторяющиеся токены вносят вклад при каждом вхождении.

Значения $x_+$ и $x_-$ **не являются вероятностями**. Это числовые признаки, характеризующие связь лексики заголовка с двумя классами.

### 5. Классификация с помощью логистической регрессии

Логистическая регрессия преобразует частотные признаки в оценку вероятности положительной тональности:

$$
z=b+\beta_+x_+ + \beta_-x_-,
$$

$$
p=P(y=1\mid h)=\frac{1}{1+e^{-z}}.
$$

Здесь $b$, $\beta_+$ и $\beta_-$ — параметры, которые подбираются при обучении.

Модель минимизирует логарифмическую функцию потерь:

$$
L=
-\frac{1}{N}
\sum_{i=1}^{N}
\left[
y_i\ln p_i+(1-y_i)\ln(1-p_i)
\right].
$$

Она штрафует модель за расхождение прогнозов с известными метками, особенно за уверенные неправильные ответы.

Для получения класса используется порог, например:

$$
\hat y=
\begin{cases}
1, & p\geq 0.5,\\
0, & p<0.5.
\end{cases}
$$

Вероятность около 0,5 означает неопределённость между двумя классами. **Она не является прогнозом нейтральной тональности:** отдельного нейтрального класса в этой модели нет.

При использовании `LogisticRegression` со свободным членом в sklearn достаточно передать два признака $x_+$ и $x_-$: константу библиотека учитывает самостоятельно.


### 6. Влияние частот и длины текста

Сырые частоты зависят от объёма обучающих данных. Если отрицательных текстов значительно больше, слова могут иметь более высокие отрицательные частоты просто из-за размера класса.

Один из вариантов нормировки:

$$
\widetilde{\operatorname{freq}}(w,c)=
\frac{\operatorname{freq}(w,c)}
{\sum_{v\in V}\operatorname{freq}(v,c)},
$$

где $V$ — словарь обучающего корпуса. Вместо абсолютного числа вхождений используется доля слова среди всех токенов класса.

Длина заголовка также влияет на признаки: в длинном тексте суммируется больше слагаемых. При необходимости вместо сумм можно использовать средние значения по токенам.

Такие нормировки являются отдельными вариантами метода. Их полезность проверяется на валидационной выборке. Масштабирование признаков перед логистической регрессией также обучается только на train.

### 7. Оценка качества

Размеченный корпус разделяется на обучающую и проверочную части **до построения словаря частот**. На обучающей части формируются признаки и обучается классификатор, после чего зафиксированная процедура применяется к проверочным текстам.

Для оценки используются:

- **precision** — насколько точны предсказания выбранного класса;
- **recall** — какую долю текстов этого класса удалось найти;
- **F1** — гармоническое среднее precision и recall;
- **macro-F1** — среднее F1 по классам с одинаковым весом каждого класса;
- **Gini** = 2 * AUC - 1.

При кросс-валидации словарь частот необходимо заново строить внутри каждой обучающей части, поскольку он зависит от меток.

### 8. Применение к новостям компаний

В нашем кейсе можно использовать две постановки.

**Анализ тональности.** Словарь частот и классификатор обучаются на внешнем корпусе с положительными и отрицательными метками. Затем модель применяется к нашим новостям. Её оценки можно агрегировать по ИНН и использовать как признаки для прогнозирования дефолта.

**Прямая классификация по дефолту.** Частоты подсчитываются по новостям дефолтных и недефолтных компаний. Получается вектор:

$$
\mathbf{x}(h)=
\left[
1,\;
\sum_{j=1}^{m}\operatorname{freq}(w_j,\text{дефолт}),\;
\sum_{j=1}^{m}\operatorname{freq}(w_j,\text{недефолт})
\right].
$$

В этом случае модель определяет связь лексики с дефолтом, а не тональность текста. Положительная новость о дефолтной компании всё равно имеет метку дефолта.

Поскольку у одной компании может быть много новостей, при оценке прогноза дефолта необходимо учитывать уровень ИНН. Компании с большим числом публикаций сильнее влияют на частоты, если каждую новость учитывать с одинаковым весом.

### Ограничения подхода

Представление из трёх чисел компактно и позволяет проследить каждый этап расчёта. При этом оно теряет порядок слов и значительную часть контекста. Разные заголовки могут получить одинаковые векторы, а частые общие слова — внести большой вклад без выраженной связи с тональностью.

Метод также не различает роли компаний внутри одной публикации: без указания целевой компании одинаковый текст получает одинаковый вектор. Поэтому такой подход используется как понятная базовая модель, с которой можно сравнивать более сложные способы анализа текста.

### Тональность и дефолт: два эксперимента с частотами слов

Мы сравним два способа получить числовую оценку новостного заголовка:

1. Обучим модель на позитивных и негативных текстах **RuSentiTweet**, затем применим её к новостям компаний.
2. Обучим такую же модель на наших новостях: **дефолт компании — условный негатив, отсутствие дефолта — условный позитив**.

В обоих случаях используем **pymorphy3**, одинаковую очистку текста и вектор из трёх чисел. TF-IDF и нейросетевых представлений здесь нет. Качество ранжирования дефолтных компаний оценим через Gini.

### Что такое RuSentiTweet

**RuSentiTweet** — корпус из **13 392 русскоязычных твитов**, размеченных вручную. Автор выделяет пять классов: `positive`, `negative`, `neutral`, `speech` и `skip`. Мы оставляем только **positive и negative** и сохраняем авторское разбиение на train и test.

Это общая тональность твитов, а не финансовые новости и не отношение к конкретной компании. Внешний test проверяет распознавание тональности, а наши новости — перенос полученного сигнала на задачу дефолта.

### Загрузка данных

Загружаем исходные CSV из репозитория автора. `sentiment_target = 1` означает позитив, `0` — негатив. Это отдельная переменная, она не совпадает по смыслу с нашим `target` дефолта.

````python
BASE_URL = 'https://raw.githubusercontent.com/sismetanin/rusentitweet/main/'

def load_external(part):
    path = DATA_DIR / f'rusentitweet_{part}.csv'
    if not path.exists():
        urllib.request.urlretrieve(BASE_URL + path.name, path)
    raw = pd.read_csv(path, dtype={'id': str})
    df = raw.loc[raw['label'].isin(['positive', 'negative'])].copy()
    assert df['text'].notna().all(), 'Во внешнем корпусе есть пустые тексты.'
    df['sentiment_target'] = df['label'].map({'negative': 0, 'positive': 1}).astype(int)
    return raw, df.reset_index(drop=True)

external_raw_train, external_train = load_external('train')
external_raw_test, external_test = load_external('test')
assert set(external_train['id']).isdisjoint(set(external_test['id']))
show_table(pd.DataFrame([
    {'Разбиение': part, 'Всего в исходнике': len(raw),
     'Позитив': int((df.label == 'positive').sum()),
     'Негатив': int((df.label == 'negative').sum()), 'Оставлено': len(df)}
    for part, raw, df in [('train', external_raw_train, external_train),
                          ('test', external_raw_test, external_test)]
]))
````

### Единая обработка текста

Приводим текст к нижнему регистру, удаляем ссылки, пользовательские `@упоминания`, HTML-теги, числа и пунктуацию. Сохраняем русские и латинские слова, заменяем `ё` на `е`, приводим слова к нормальной форме с помощью **pymorphy3**.

Удаляем русские стоп-слова NLTK и их нормальные формы. **«Не», «ни», «нет», «без» сохраняем**, поскольку отрицание важно для тональности. Это всё ещё модель «мешка слов»: порядок слов и связь отрицания с конкретным словом она не учитывает. Для неоднозначных слов используем первый разбор pymorphy3; это упрощение без учёта контекста.

Одна и та же функция применяется к твитам и заголовкам. Результаты разбора слов и обработки текстов кешируются.

````python
nltk.download('stopwords', quiet=True)
morph = pymorphy3.MorphAnalyzer()

@lru_cache(maxsize=200_000)
def lemma(word):
    return morph.parse(word)[0].normal_form.replace('ё', 'е')

NEGATIONS = {'не', 'ни', 'нет', 'без'}
raw_stops = {w.replace('ё', 'е') for w in stopwords.words('russian')} - NEGATIONS
STOP_WORDS = (raw_stops | {lemma(w) for w in raw_stops}) - NEGATIONS
TOKEN_RE = re.compile(r'[а-яa-z]+')

@lru_cache(maxsize=200_000)
def preprocess(text):
    text = unescape(str(text)).lower().replace('ё', 'е')
    text = re.sub(r'https?://\S+|www\.\S+|@[\w]+', ' ', text)
    text = re.sub(r'<[^>]+>', ' ', text)
    tokens = []
    for word in TOKEN_RE.findall(text):
        if word in STOP_WORDS:
            continue
        normal = lemma(word)
        if normal not in STOP_WORDS:
            tokens.append(normal)
    return tuple(tokens)

for df in [external_train, external_test]:
    df['tokens'] = df['text'].map(preprocess)

show_table(pd.DataFrame({
    'Исходный текст': ['Компания не выплатила долг кредиторам', 'Компания увеличила прибыль'],
    'После обработки': [' '.join(preprocess(t)) for t in
                        ['Компания не выплатила долг кредиторам', 'Компания увеличила прибыль']]
}), widths=[55, 45])
````

### От частот к вектору длины три

По **обучающим текстам** считаем, сколько раз каждое слово встретилось в позитивном и негативном классах. Повторные вхождения слова учитываем:

$$
f_{+}(w)=\sum_{i:y_i=1}\operatorname{count}(w,d_i),\qquad
f_{-}(w)=\sum_{i:y_i=0}\operatorname{count}(w,d_i).
$$

Для текста $d$ суммируем соответствующие частоты всех его слов:

$$
x(d)=\left[1,\ s_{+}(d),\ s_{-}(d)\right],\qquad
s_{+}(d)=\sum_{w\in d}f_{+}(w),\quad
s_{-}(d)=\sum_{w\in d}f_{-}(w).
$$

Например, если после обработки осталось три слова, каждое из них добавляет свою частоту в позитивную и негативную сумму. Неизвестные слова добавляют ноль. При пустом тексте или полностью неизвестной лексике получаем $[1,0,0]$.

Обучаем **логистическую регрессию**. Первая координата — константа для свободного члена. В коде передаём две частотные суммы, а единицу учитывает `fit_intercept=True`. Перед обучением стандартизируем две суммы с помощью `StandardScaler`, обученного только на train:

$$
P(\text{позитив}\mid d)=\sigma\left(b+\beta_{+}z_{+}(d)+\beta_{-}z_{-}(d)\right),\qquad
\sigma(t)=\frac{1}{1+e^{-t}}.
$$

Сырые частоты зависят от размеров классов и длины текстов. Здесь это часть простого учебного подхода: балансировку и TF-IDF не применяем.

````python
def build_frequencies(documents, positive_mask):
    positive, negative = Counter(), Counter()
    for tokens, is_positive in zip(documents, positive_mask):
        (positive if is_positive else negative).update(tokens)
    return positive, negative


def vectorize(documents, frequencies):
    positive, negative = frequencies
    return np.array([
        [sum(positive[w] for w in tokens), sum(negative[w] for w in tokens)]
        for tokens in documents
    ], dtype=np.float64).reshape(-1, 2)


def new_model():
    return make_pipeline(
        StandardScaler(),
        LogisticRegression(C=1.0, fit_intercept=True, max_iter=2000, random_state=SEED)
    )


external_frequencies = build_frequencies(
    external_train['tokens'], external_train['sentiment_target'].eq(1)
)
X_external_train = vectorize(external_train['tokens'], external_frequencies)
X_external_test = vectorize(external_test['tokens'], external_frequencies)
external_model = new_model()
external_model.fit(X_external_train, external_train['sentiment_target'])

# Здесь вероятность класса 1 — вероятность ПОЗИТИВА.
external_test_p_positive = external_model.predict_proba(X_external_test)[:, 1]
show_table(pd.DataFrame([{
    'Проверка': 'RuSentiTweet: авторский test',
    'ROC AUC (позитив = 1)': roc_auc_score(external_test.sentiment_target, external_test_p_positive),
    'Accuracy (порог 0.5)': accuracy_score(external_test.sentiment_target, external_test_p_positive >= 0.5)
}]), widths=[34, 22, 22, 22])
````

### Загружаем две новостные выборки

Данные агрегатора объединены с флагом дефолта компании. **Единица строки — пара «публикация — компания», ключ `(news_id, inn)`**. Один заголовок может повторяться для нескольких компаний.

- `all_mentions`: компания с указанным ИНН упомянута в новости, независимо от её роли.
- `main_role`: компания с указанным ИНН — главный участник новости (`is_main_role = True`). Это подмножество первой выборки, а не независимый test.


````python
# Можно указать абсолютные пути, например к файлам на подключённом Google Drive.
SAMPLE_FILES = {
    'all_mentions': 'train_headlines_all_mentions.csv',
    'main_role': 'train_headlines_main_role.csv',
}
SAMPLE_LABELS = {'all_mentions': 'Все упоминания', 'main_role': 'Главная роль'}


def find_sample(filename):
    path = Path(filename)
    candidates = [path, ROOT / path.name, DATA_DIR / path.name]
    if path.suffix != '.gz':
        candidates += [Path(str(p) + '.gz') for p in list(candidates)]
    return next((p for p in candidates if p.is_file()), None)


missing = [name for name in SAMPLE_FILES.values() if find_sample(name) is None]
if missing:
    if IN_COLAB:
        from google.colab import files
        print('Загрузите файлы:', ', '.join(missing))
        files.upload()  # Ожидает завершения загрузки.
    else:
        raise FileNotFoundError('Укажите пути в SAMPLE_FILES: ' + ', '.join(missing))


def read_news(filename):
    path = find_sample(filename)
    if path is None:
        raise FileNotFoundError(f'Файл {filename} не найден. Проверьте имя и завершение загрузки.')
    df = pd.read_csv(path, encoding='utf-8-sig', dtype={'inn': str, 'news_id': str})
    if 'target' not in df.columns and 'score_3' in df.columns:
        df = df.rename(columns={'score_3': 'target'})
    elif {'target', 'score_3'}.issubset(df.columns):
        assert pd.to_numeric(df.target).equals(pd.to_numeric(df.score_3)), 'target и score_3 противоречат друг другу.'
        df = df.drop(columns='score_3')
    required = {'news_id', 'inn', 'title', 'target', 'is_main_role', 'split'}
    assert required.issubset(df.columns), f'Нет полей: {required - set(df.columns)}'
    assert df[['news_id', 'inn', 'title', 'target']].notna().all().all(), 'Есть пропуски в обязательных полях.'
    df['inn'] = df['inn'].str.strip()
    assert df.inn.str.fullmatch(r'\d{10}|\d{12}').all(), 'Проверьте формат ИНН.'
    df['target'] = pd.to_numeric(df['target'], errors='raise')
    assert set(df.target.unique()) == {0, 1}, 'Ожидаем оба класса: 0 и 1.'
    df['target'] = df['target'].astype(int)
    assert not df.duplicated(['news_id', 'inn']).any(), 'Повторяется ключ (news_id, inn).'
    assert df.groupby('inn')['target'].nunique().max() == 1, 'Несогласованный target внутри ИНН.'
    assert df['split'].eq('train').all(), 'Ожидается обучающая выборка.'
    return df


samples = {key: read_news(filename) for key, filename in SAMPLE_FILES.items()}
main_flags = samples['main_role']['is_main_role'].astype(str).str.lower()
assert main_flags.isin(['true', '1']).all(), 'В main_role есть неглавные упоминания.'
all_index = samples['all_mentions'].set_index(['news_id', 'inn'])
main_index = samples['main_role'].set_index(['news_id', 'inn'])
assert main_index.index.isin(all_index.index).all(), 'main_role не является подмножеством all_mentions.'
assert main_index[['title', 'target']].equals(all_index.loc[main_index.index, ['title', 'target']])

composition = []
for key, df in samples.items():
    companies = df.drop_duplicates('inn')
    composition.append({
        'Выборка': SAMPLE_LABELS[key], 'Строк': len(df),
        'Строк target=1': int(df.target.sum()), 'Доля target=1, %': df.target.mean() * 100,
        'Компаний': len(companies), 'Дефолтных компаний': int(companies.target.sum())
    })
show_table(pd.DataFrame(composition), widths=[24, 12, 16, 18, 14, 16])
````

### Скоринг заголовков и направление Gini

Применяем готовые частоты и модель RuSentiTweet к заголовкам. **На новостях внешнюю модель не дообучаем.** Для ранжирования дефолтов используем вероятность негатива:

$$
\text{score}_{external}(d)=1-P(\text{позитив}\mid d).
$$

Считаем метрику относительно `target = 1` — дефолта:

$$
Gini=2\cdot ROC\ AUC-1.
$$

Gini около нуля означает отсутствие ранжирующего сигнала, отрицательное значение — обратное направление. **Знак не переворачиваем по результату и модуль не берём.** Вероятность негатива не является вероятностью дефолта.

Основная оценка — по строкам, то есть по парам «публикация — компания». Дополнительно усредняем скоры всех заголовков по ИНН и считаем Gini по компаниям. В первой оценке компании с большим числом новостей имеют больший вес, во второй каждая компания представлена один раз.

````python
def coverage(tokens, frequencies):
    positive, negative = frequencies
    return sum(w in positive or w in negative for w in tokens) / len(tokens) if tokens else 0.0


def metric_row(df, score_column, sample, method, evaluation):
    company = df.groupby('inn', as_index=False).agg(
        target=('target', 'first'), score=(score_column, 'mean')
    )
    auc_rows = roc_auc_score(df.target, df[score_column])
    auc_companies = roc_auc_score(company.target, company.score)
    return {
        'Выборка': SAMPLE_LABELS[sample], 'Подход': method, 'Оценка': evaluation,
        'AUC строк': auc_rows, 'Gini строк': 2 * auc_rows - 1,
        'Gini компаний': 2 * auc_companies - 1,
    }


results = []
for key, df in samples.items():
    df['tokens'] = df['title'].map(preprocess)
    df['title_lemma'] = df['tokens'].map(' '.join)
    X_news = vectorize(df.tokens, external_frequencies)
    df['ext_freq_positive'] = X_news[:, 0]
    df['ext_freq_negative'] = X_news[:, 1]
    df['external_p_negative'] = 1 - external_model.predict_proba(X_news)[:, 1]
    df['external_vocab_coverage'] = df.tokens.map(lambda t: coverage(t, external_frequencies))
    results.append(metric_row(df, 'external_p_negative', key, 'RuSentiTweet', 'Перенос на news train'))

show_table(pd.DataFrame(results), widths=[18, 16, 24, 14, 14, 14])
show_table(pd.DataFrame([
    {'Выборка': SAMPLE_LABELS[key],
     'Пустых после очистки': int(df.tokens.map(len).eq(0).sum()),
     'Среднее покрытие словаря, %': 100 * df.external_vocab_coverage.mean(),
     'Заголовков без известных слов': int(df.external_vocab_coverage.eq(0).sum())}
    for key, df in samples.items()
]), widths=[25, 23, 27, 25])
````

### Обучение частот и модели на флаге дефолта

Теперь строим словари по нашим заголовкам. `target = 0` — условный позитив, `target = 1` — условный негатив. Частота слова считается **по строкам выборки**: заголовок про несколько компаний участвует несколько раз, с меткой каждой компании.

$$
f_{+}^{news}(w)=\sum_{i:target_i=0}\operatorname{count}(w,d_i),\qquad
f_{-}^{news}(w)=\sum_{i:target_i=1}\operatorname{count}(w,d_i).
$$

Вектор остаётся тем же: $[1,s_{+},s_{-}]$. Логистическую регрессию теперь обучаем сразу предсказывать `target = 1`, поэтому её выход направлен в сторону дефолта.

Это **классификация по метке дефолта**, а не самостоятельная разметка тональности. Модель ищет слова, связанные с дефолтными компаниями, в том числе отраслевые слова и названия компаний. ИНН и выручка в признаки не входят.

Для каждого из двух сэмплов обучаем **собственные частоты и собственную модель**. Сначала считаем запрошенный Gini на train. Затем получаем out-of-fold (OOF) оценку: делим **компании** на пять частей со стратификацией по дефолту. Для каждой части заново строим частоты, scaler и модель только на остальных компаниях. Предсказываем заголовки исключённых компаний.

**Почему нужны обе оценки:** train Gini использует те же метки, по которым построены частоты и обучена модель, и потому оптимистичен. OOF уменьшает этот эффект. Одинаковые публикации, относящиеся к разным компаниям, сохраняем: они могут попасть в разные folds. Поэтому OOF по ИНН не гарантирует независимости текстов и не заменяет временной тест.

Ниже сохраняем **все пять значений Gini**, а не только объединённый OOF: для каждой итерации — Gini на четырёх обучающих частях и на одной проверочной. Это 5-fold CV со стратификацией на уровне уникальных ИНН; случайное разбиение отдельных новостных строк не используем.

````python
def fit_default_model(df):
    frequencies = build_frequencies(df.tokens, df.target.eq(0))
    X = vectorize(df.tokens, frequencies)
    model = new_model()
    model.fit(X, df.target)
    return frequencies, model


def score_default_model(df, frequencies, model):
    return model.predict_proba(vectorize(df.tokens, frequencies))[:, 1]


def grouped_oof(df, n_splits=5):
    companies = df[['inn', 'target']].drop_duplicates('inn').sort_values('inn').reset_index(drop=True)
    assert companies.target.value_counts().min() >= n_splits
    splitter = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=SEED)
    predictions = np.full(len(df), np.nan)
    folds = np.zeros(len(df), dtype=int)
    diagnostics = []
    for fold, (company_train, company_valid) in enumerate(
        splitter.split(companies.inn, companies.target), start=1
    ):
        train_inns = set(companies.iloc[company_train].inn)
        valid_inns = set(companies.iloc[company_valid].inn)
        assert train_inns.isdisjoint(valid_inns)
        train_mask = df.inn.isin(train_inns).to_numpy()
        valid_mask = df.inn.isin(valid_inns).to_numpy()
        train_part, valid_part = df.loc[train_mask], df.loc[valid_mask]
        frequencies, model = fit_default_model(train_part)
        predictions[valid_mask] = score_default_model(valid_part, frequencies, model)
        train_scores = score_default_model(train_part, frequencies, model)
        valid_scores = predictions[valid_mask]
        valid_companies = valid_part[['inn', 'target']].assign(score=valid_scores).groupby('inn').agg(
            target=('target', 'first'), score=('score', 'mean')
        )
        folds[valid_mask] = fold
        shared_news = set(train_part.news_id) & set(valid_part.news_id)
        diagnostics.append({
            'Fold': fold, 'Компаний train': len(train_inns), 'Компаний valid': len(valid_inns),
            'Строк train': len(train_part), 'Строк valid': len(valid_part),
            'Дефолтных компаний valid': int(valid_companies.target.sum()),
            'Доля дефолтов valid: компании, %': 100 * valid_companies.target.mean(),
            'Доля дефолтов valid: строки, %': 100 * valid_part.target.mean(),
            'Gini train: строки': 2 * roc_auc_score(train_part.target, train_scores) - 1,
            'Gini valid: строки': 2 * roc_auc_score(valid_part.target, valid_scores) - 1,
            'Gini valid: компании': 2 * roc_auc_score(valid_companies.target, valid_companies.score) - 1,
            'Общих news_id': len(shared_news),
            'Доля строк valid с общим news_id, %': 100 * valid_part.news_id.isin(shared_news).mean()
        })
    assert np.isfinite(predictions).all() and (folds > 0).all()
    return predictions, folds, diagnostics


news_models = {}
oof_diagnostics = []
# Позволяет повторно запускать ячейку без дублирования строк в отчёте.
results = [r for r in results if r['Подход'] == 'RuSentiTweet']
for key, df in samples.items():
    frequencies, model = fit_default_model(df)
    news_models[key] = (frequencies, model)
    X = vectorize(df.tokens, frequencies)
    df['news_freq_nondefault'] = X[:, 0]
    df['news_freq_default'] = X[:, 1]
    df['default_score_train'] = model.predict_proba(X)[:, 1]
    df['default_score_oof'], df['oof_fold'], diagnostics = grouped_oof(df)
    oof_diagnostics.extend([{'Выборка': SAMPLE_LABELS[key], **r} for r in diagnostics])
    results.append(metric_row(df, 'default_score_train', key, 'Метка дефолта', 'На обучающих строках'))
    results.append(metric_row(df, 'default_score_oof', key, 'Метка дефолта', 'OOF по ИНН, 5 folds'))
    print(f'{SAMPLE_LABELS[key]}: обучение и OOF завершены.')
````

Для каждого сэмпла показываем все пять итераций. **Train** — четыре части, на которых в этой итерации построены частоты и обучена модель. **Valid** — оставшаяся часть компаний. Это части исходного train-файла, а не наш отдельный test.

Стратификация сохраняет долю дефолтов **среди компаний**. Доля дефолтных **строк** может значительно меняться, поскольку число новостей у компаний разное. Поэтому рядом с Gini приводим объём и состав проверочной части.

Средний Gini пяти folds и Gini объединённых OOF-прогнозов — разные величины: объединённая оценка дополнительно сравнивает скоры, полученные разными моделями для разных folds. Она не обязана совпадать со средним, даже взвешенным по числу строк.

### Два уровня моделирования

На уровне **новости** можно определить тональность, тему, участников и наличие риск-события. На уровне **компании на дату оценки** — агрегировать информацию из допустимого временного окна и построить прогноз.

**Оценка негативности новости не равна вероятности дефолта компании.** Даже если обе представлены числом от нуля до единицы, они относятся к разным событиям и объектам.

Из новостного потока можно сформировать, например, частоту определённых событий, долю публикаций по теме, среднюю негативность, изменение информационного фона или векторное представление содержания. Затем нужно проверить, есть ли в этих признаках информация о будущем дефолте.

---

Рассматриваются четыре подхода: **TF-IDF, LDA, NMF и Word2Vec**. TF-IDF описывает документ весами слов и словосочетаний. LDA и NMF выделяют повторяющиеся сочетания слов, которые можно интерпретировать как темы. Word2Vec представляет слова векторами, близость которых связана с употреблением в похожем окружении. Во всех четырёх случаях прогноз дефолта строится с помощью **логистической регрессии**.

### Данные и задача

Данные новостного агрегатора объединены с флагом дефолта компании. **Единица строки — пара «публикация — компания», определяемая ключом `(news_id, inn)`.** Одна публикация может встречаться в нескольких строках, если в ней упомянуто несколько компаний.

В файле `train_texts_all_mentions.csv` компания может быть любым участником новости. В `train_texts_main_role.csv` она выступает главным участником согласно полю `is_main_role`. Первый файл содержит 45 547 строк и 1 200 компаний, из которых 300 имеют флаг дефолта. Во втором — 16 716 строк и 615 компаний, из которых 138 имеют флаг дефолта. Главная роль означает значимость компании в публикации; она не сообщает, является компания, например, истцом или ответчиком.

**Объект прогнозирования — компания.** Каждую её новость представим вектором, затем усредним векторы и получим одну строку признаков на ИНН. Метка `target=1` означает дефолт компании. Она не означает, что каждая новость об этой компании имеет негативную тональность.

Для сравнения методов используется пятикратная кросс-валидация по компаниям. Обе выборки сохраняют исходный состав; отдельная тестовая выборка в этих экспериментах не участвует.

````python
#@title Читаем CSV и проверяем ключи
def read_news(path):
    path = Path(path)
    if not path.is_file():
        raise FileNotFoundError(f'Не найден {path}. Загрузите распакованный CSV, дождитесь '
                                'завершения загрузки и проверьте SAMPLE_FILES.')
    needed = {'news_id','inn','text','target','score_3','is_main_role',
              'published_at','company_name','revenue','split'}
    df = pd.read_csv(path, encoding='utf-8-sig', dtype={'inn':str,'news_id':str},
                     usecols=lambda c: c in needed)
    if 'target' not in df and 'score_3' in df:
        df = df.rename(columns={'score_3':'target'})
    elif {'target','score_3'} <= set(df):
        if not pd.to_numeric(df.target).equals(pd.to_numeric(df.score_3)):
            raise ValueError('target и score_3 противоречат друг другу.')
        df = df.drop(columns='score_3')
    required = {'news_id','inn','text','target','is_main_role'}
    if not required <= set(df):
        raise ValueError(f'Не хватает колонок: {required-set(df)}')
    if df[['news_id','inn','target']].isna().any().any():
        raise ValueError('Пропуск в ключе или target.')
    df['inn'] = df.inn.str.strip()
    if not df.inn.str.fullmatch(r'(?:\d{10}|\d{12})').all():
        raise ValueError('ИНН должен содержать 10 или 12 цифр.')
    df['target'] = pd.to_numeric(df.target, errors='raise')
    if set(df.target.unique()) != {0,1}:
        raise ValueError('Ожидаются оба класса: 0 и 1.')
    df['target'] = df.target.astype('int64')
    if df.groupby('inn').target.nunique().max() != 1:
        raise ValueError('У одной компании разные target.')
    if df.duplicated(['news_id','inn']).any():
        raise ValueError('Повторяется ключ (news_id, inn): проверьте версии публикаций.')
    flags = df.is_main_role.astype(str).str.strip().str.lower()
    if not flags.isin(['true','false','1','0']).all():
        raise ValueError('Неизвестное значение is_main_role.')
    df['is_main_role'] = flags.isin(['true','1'])
    if 'split' in df and not df.split.eq('train').all():
        raise ValueError('Здесь ожидается исходный train без test/hold.')
    df['text'] = df.text.fillna('').astype(str)
    return df

samples_raw = {key:read_news(path) for key,path in SAMPLE_FILES.items()}
assert samples_raw['main_role'].is_main_role.all()
reference = samples_raw['all_mentions'].set_index(['news_id','inn'])
main_index = samples_raw['main_role'].set_index(['news_id','inn'])
assert main_index.index.isin(reference.index).all(), 'main_role не является подмножеством all_mentions.'
assert main_index[['text','target']].equals(reference.loc[main_index.index,['text','target']])
composition = []
for key,df in samples_raw.items():
    firms = df.drop_duplicates('inn')
    composition.append({'Срез':SAMPLE_LABELS[key], 'Строк':len(df), 'Компаний':len(firms),
                        'Дефолтов компаний':int(firms.target.sum()),
                        'DR компаний, %':100*firms.target.mean(),
                        'Пустых текстов':int(df.text.str.strip().eq('').sum())})
show_table(pd.DataFrame(composition), [24,13,14,18,17,14])
pd.DataFrame(composition).to_csv(RESULT_DIR/'source_composition.csv', index=False, encoding='utf-8-sig')
del reference, main_index
````

### Разбиение по компаниям

Если случайно разделить строки с новостями, публикации об одной компании могут оказаться и в обучении, и в валидации. Модель сможет узнавать компанию по названию или характерным словам. Такая проверка не отвечает на вопрос, насколько хорошо модель работает на других компаниях.

Поэтому сначала составляется таблица уникальных ИНН, а затем компании распределяются по пяти фолдам. Все новости одного ИНН относятся к одному фолду. В каждом из пяти экспериментов четыре фолда используются для обучения, пятый — для проверки. В результате каждая компания получает один прогноз от модели, которая не обучалась на её строках. Это **out-of-fold прогноз**, или OOF.

У компаний, представленных в обоих срезах, номер фолда одинаков. При разбиении учитываются флаг дефолта и наличие компании в срезе главных участников. Так соотношение классов сохраняется приблизительно одинаковым между фолдами внутри каждого среза. Все 1 200 компаний остаются в первом срезе, все 615 — во втором.

Словарь, IDF, тематические модели и Word2Vec обучаются заново на обучающей части каждого фолда. То же относится к масштабированию признаков и логистической регрессии. Лемматизацию можно выполнить до разбиения: она использует готовый морфологический словарь и не оценивает параметры по нашему корпусу.

Разбиение по компаниям не исключает общих публикаций. Например, статья о споре двух компаний может оказаться в обучении для истца и в валидации для ответчика. Такие строки сохраняются, а пересечение публикаций измеряется отдельно. Полученная оценка характеризует прогноз для отложенных компаний при возможном пересечении новостного содержания. Для проверки прогноза на будущих событиях необходимо дополнительно разделять данные по времени с учётом дат дефолта.

````python
#@title Пять фолдов по компаниям
samples = {key: df.sort_values(['inn','news_id']).reset_index(drop=True)
           for key, df in samples_raw.items()}
company_union = pd.concat([d[['inn','target']] for d in samples.values()],ignore_index=True)
assert company_union.groupby('inn').target.nunique().max() == 1
company_union = company_union.drop_duplicates('inn').sort_values('inn').reset_index(drop=True)
company_union['in_main_role'] = company_union.inn.isin(samples['main_role'].inn)
strata = company_union.target.astype(str) + '_' + company_union.in_main_role.astype(str)
if strata.value_counts().min() < N_SPLITS:
    raise ValueError('В одной из групп target × наличие главной роли меньше пяти компаний.')
company_union['fold'] = -1
splitter = StratifiedKFold(n_splits=N_SPLITS,shuffle=True,random_state=SEED)
for fold,(_,idx) in enumerate(splitter.split(company_union,strata),1):
    company_union.loc[idx,'fold'] = fold
fold_map = company_union.set_index('inn')['fold']
for key,df in samples.items():
    df['fold'] = df.inn.map(fold_map).astype(int)
    assert len(df) == len(samples_raw[key])
    firm = df.drop_duplicates('inn')
    for fold in range(1,N_SPLITS+1):
        if firm.loc[firm.fold.eq(fold),'target'].nunique() != 2:
            raise ValueError(f'{key}, fold={fold}: в валидации нет одного из классов.')
company_union.to_csv(RESULT_DIR/'company_folds.csv',index=False,encoding='utf-8-sig')
fold_composition = pd.concat([
    d.drop_duplicates('inn').groupby('fold').target.agg(Компаний='size',Дефолтов='sum')
     .reset_index().assign(Срез=SAMPLE_LABELS[k]) for k,d in samples.items()
],ignore_index=True)
fold_composition['DR, %'] = 100.0*fold_composition['Дефолтов']/fold_composition['Компаний']
show_table(fold_composition[['Срез','fold','Компаний','Дефолтов','DR, %']]
           .rename(columns={'fold':'Фолд'}),[32,12,18,18,20])
del samples_raw
gc.collect()
````

### Подготовка текста

Слова «кредит», «кредита» и «кредитами» обозначают одно понятие, но без обработки занимают разные столбцы словаря. **Лемматизация** приводит словоформы к начальной форме: «кредита» → «кредит», «выплатила» → «выплатить». Для русского языка используется pymorphy3.

Из текста удаляются HTML-разметка и ссылки, буквы переводятся в нижний регистр, «ё» заменяется на «е». Затем выделяются последовательности русских и латинских букв. Числа и знаки препинания в это представление не входят. Поэтому сумму долга или срок просрочки такая модель непосредственно не сохраняет: их можно извлекать в отдельные признаки.

После лемматизации удаляются стоп-слова — частые служебные слова, например «и», «в», «на». Исключение сделано для «не», «ни», «нет» и «без»: отрицание может менять смысл события. Сравним «компания выплатила долг» и «компания не выплатила долг». Удаление «не» сделало бы их представления одинаковыми.

Фраза «Компания не выплатила долги кредиторам» принимает вид, близкий к «компания не выплатить долг кредитор». При этом грамматика и часть контекста теряются. Pymorphy3 выбирает разбор отдельного слова, поэтому омонимы и названия организаций могут обрабатываться неоднозначно.

Один и тот же текст достаточно лемматизировать один раз. Результат затем связывается со всеми строками, в которых встречается публикация. Это экономит вычисления и сохраняет пары «публикация — компания».

````python
#@title pymorphy3, стоп-слова и пример
if not nltk.download('stopwords',quiet=True):
    raise RuntimeError('Не удалось загрузить стоп-слова. Проверьте интернет.')
morph = pymorphy3.MorphAnalyzer()
@lru_cache(maxsize=None)
def lemma(word):
    return morph.parse(word)[0].normal_form.replace('ё','е')
NEGATIONS = {'не','ни','нет','без'}
raw_stops = {w.replace('ё','е') for w in stopwords.words('russian')} - NEGATIONS
STOP_WORDS = (raw_stops | {lemma(w) for w in raw_stops}) - NEGATIONS
TOKEN_RE = re.compile(r'[а-яa-z]+')
def clean_text(text):
    text = unescape(str(text)).replace('\\n','\n').replace('\\t',' ')
    text = re.sub(r'<[^>]+>',' ',text)
    text = re.sub(r'https?://\S+|www\.\S+',' ',text)
    return re.sub(r'\s+',' ',text).strip()
def preprocess(text):
    tokens = []
    for word in TOKEN_RE.findall(clean_text(text).lower().replace('ё','е')):
        if word in STOP_WORDS:
            continue
        normal = lemma(word)
        if normal not in STOP_WORDS:
            tokens.append(normal)
    return ' '.join(tokens)
demo_texts = ['Компания не выплатила долги кредиторам.',
              'Компания увеличила выручку и открыла новый завод.',
              'Кредиторы согласовали перенос платежей.']
show_table(pd.DataFrame({'Исходный текст':demo_texts,
                         'Леммы':[preprocess(t) for t in demo_texts]}),[50,50])
````

````python
#@title Обрабатываем уникальные тексты и сохраняем кэш
pool = pd.concat([d[['news_id','text']] for d in samples.values()],ignore_index=True)
assert pool.groupby('news_id').text.nunique().max() == 1, 'Разные версии одного news_id.'
pool = pool.drop_duplicates('news_id').reset_index(drop=True)
pool['text_hash'] = pool.text.map(lambda t:hashlib.sha256(clean_text(t).encode()).hexdigest())
unique = pool.drop_duplicates('text_hash')
signature = json.dumps({'version':'day2-all-data-v2',
                        'pymorphy3':versions['pymorphy3'],'stops':sorted(STOP_WORDS)},sort_keys=True)
cache_path = CACHE_DIR/f'lemmas_{hashlib.sha256(signature.encode()).hexdigest()[:16]}.json'
cache = json.loads(cache_path.read_text()) if cache_path.exists() else {}
missing = unique.loc[~unique.text_hash.isin(cache)]
for n,row in enumerate(tqdm(missing.itertuples(index=False),total=len(missing),desc='Лемматизация'),1):
    cache[row.text_hash] = preprocess(row.text)
    if n % 1000 == 0:
        cache_path.write_text(json.dumps(cache,ensure_ascii=False))
cache_path.write_text(json.dumps(cache,ensure_ascii=False))
pool['lemma_text'] = pool.text_hash.map(cache)
lookup = pool.set_index('news_id')
for df in samples.values():
    df['lemma_text'] = df.news_id.map(lookup.lemma_text)
    df['text_hash'] = df.news_id.map(lookup.text_hash)
show_table(pd.DataFrame([
    {'Срез':SAMPLE_LABELS[k],'Строк':len(d),'Уникальных текстов':d.text_hash.nunique(),
     'Пустых после очистки':int(d.lemma_text.eq('').sum()),
     'Среднее число лемм':d.lemma_text.str.split().str.len().mean()}
    for k,d in samples.items()]),[25,13,20,22,20])
del pool,unique,lookup,missing,cache
gc.collect()
````

### От новости к компании

Пусть у компании $i$ есть $n_i$ новостей с векторами $z_{i1},\ldots,z_{in_i}$. Её текстовое представление определяется средним:

$$\bar z_i=\frac{1}{n_i}\sum_{j=1}^{n_i}z_{ij}.$$

Например, если две новости представлены долями тем $(0.8,0.2)$ и $(0.4,0.6)$, среднее представление компании равно $(0.6,0.4)$. Каждая новость вносит одинаковый вклад независимо от длины. При этом компания с сотней публикаций, как и компания с одной публикацией, занимает одну строку в модели дефолта.

Документы без известных слов дают нулевые векторы и не участвуют в среднем. Если у компании все векторы нулевые, она сохраняется с нулевыми текстовыми признаками. Для TF-IDF и Word2Vec среднее дополнительно нормируется по евклидовой длине, для тематических представлений — по сумме компонентов. Плотные признаки стандартизируются по средним и стандартным отклонениям обучающих компаний.

Усреднение описывает общее содержание новостей, но может ослабить редкий важный сигнал. Одна публикация о просрочке среди ста нейтральных сообщений будет иметь небольшой вес. Это свойство агрегации, которое следует учитывать при интерпретации результатов.

### Логистическая регрессия

Для вектора признаков компании $x_i$ модель вычисляет линейный скор и преобразует его в число от 0 до 1:

$$s_i=b+\sum_{k=1}^{p}\beta_kx_{ik},\qquad
\hat p_i=\frac{1}{1+\exp(-s_i)}.$$

Положительный коэффициент увеличивает скор при росте соответствующего признака, отрицательный — уменьшает. Коэффициенты подбираются по меткам дефолта. L2-регуляризация уменьшает чрезмерно большие коэффициенты и помогает при большом числе коррелирующих признаков. Во всех четырёх моделях используется одинаковое значение параметра регуляризации $C=1$.

### Оценка качества

ROC AUC можно интерпретировать как вероятность того, что случайная дефолтная компания получит более высокий скор, чем случайная недефолтная. При совпадении скоров учитывается половина успеха. Коэффициент Gini связан с ROC AUC формулой:

$$Gini=2\,ROC\ AUC-1.$$

Например, AUC 0.70 соответствует Gini 0.40. У случайного ранжирования Gini в среднем равен нулю. Отрицательное значение означает, что на этой выборке порядок скоров оказался хуже случайного. Менять направление скора после просмотра валидационных меток было бы подгонкой под результат проверки.

Для каждого метода рассчитываются пять значений Gini на валидации, их среднее и стандартное отклонение. Большой разброс означает чувствительность результата к составу обучающих и проверочных компаний. Стандартное отклонение пяти значений само по себе не является доверительным интервалом.

Gini на объединённых OOF-прогнозах может отличаться от среднего Gini фолдов: объединённая оценка сравнивает в том числе скоры, полученные разными моделями. Дополнительно рассчитывается Average Precision — характеристика зависимости precision от recall. Её следует рассматривать с учётом доли дефолтов.

Доля дефолтов в выборках задана при их формировании. Поэтому хороший Gini подтверждает способность ранжировать компании, но не доказывает, что значение 0.20 соответствует 20-процентной вероятности дефолта в банковском портфеле.

````python
#@title Общие функции: из новостей в компании, LR и метрики
def aggregate_companies(df,news_vectors,norm_kind):
    firms = df[['inn','target']].drop_duplicates('inn').sort_values('inn').reset_index(drop=True)
    codes = pd.Index(firms.inn).get_indexer(df.inn)
    present = (np.asarray(news_vectors.getnnz(axis=1)).ravel()>0 if sparse.issparse(news_vectors)
               else np.linalg.norm(news_vectors,axis=1)>0)
    counts = np.bincount(codes,weights=present,minlength=len(firms))
    weights = present.astype(float)/np.maximum(counts[codes],1)
    averaging = sparse.csr_matrix((weights,(codes,np.arange(len(df)))),shape=(len(firms),len(df)))
    X = normalize(averaging @ news_vectors,norm=norm_kind,axis=1)
    firms['n_news_used'] = counts.astype(int)
    return firms,X

def metrics(y,p):
    auc = roc_auc_score(y,p)
    return {'auc':auc,'gini':2*auc-1,'ap':average_precision_score(y,p)}

fold_results,prediction_frames,summaries,artifacts = [],[],[],{}
METHOD_LABELS = {'tfidf':'TF-IDF','lda':'LDA','nmf':'NMF','w2v':'Word2Vec'}

def evaluate_method(method,builder):
    # Повторный запуск заменяет результаты этого метода.
    fold_results[:] = [r for r in fold_results if r['method']!=method]
    prediction_frames[:] = [d for d in prediction_frames if d.method.iloc[0]!=method]
    summaries[:] = [r for r in summaries if r['method']!=method]
    for key,df in samples.items():
        current = []
        for fold in range(1,N_SPLITS+1):
            started = time.perf_counter()
            train = df.loc[df.fold.ne(fold)].reset_index(drop=True)
            valid = df.loc[df.fold.eq(fold)].reset_index(drop=True)
            assert set(train.inn).isdisjoint(valid.inn)
            # Обучаем представление на уникальных текстах, не умножая
            # вклад одной публикации числом упомянутых в ней компаний.
            fit_docs = train.loc[train.lemma_text.ne('')].drop_duplicates('text_hash').lemma_text.tolist()
            if not fit_docs:
                raise ValueError('Нет непустых обучающих текстов.')
            train_news,valid_news,art = builder(fit_docs,train.lemma_text,valid.lemma_text)
            norm_kind = 'l1' if method in {'lda','nmf'} else 'l2'
            firms_tr,X_tr = aggregate_companies(train,train_news,norm_kind)
            firms_va,X_va = aggregate_companies(valid,valid_news,norm_kind)
            scaler = None
            if not sparse.issparse(X_tr):
                scaler = StandardScaler()
                X_tr = scaler.fit_transform(X_tr)
                X_va = scaler.transform(X_va)
            clf = LogisticRegression(C=LR_C,solver='liblinear',max_iter=3000,
                                      class_weight=None,random_state=SEED)
            clf.fit(X_tr,firms_tr.target)
            p_tr,p_va = clf.predict_proba(X_tr)[:,1],clf.predict_proba(X_va)[:,1]
            mt,mv = metrics(firms_tr.target,p_tr),metrics(firms_va.target,p_va)
            elapsed = time.perf_counter()-started
            fold_results.append({'sample':key,'method':method,'fold':fold,
                'train_companies':len(firms_tr),'valid_companies':len(firms_va),
                'train_gini':mt['gini'],'valid_gini':mv['gini'],'valid_auc':mv['auc'],
                'valid_ap':mv['ap'],'valid_no_text':int(firms_va.n_news_used.eq(0).sum()),
                'seconds':elapsed})
            pred = firms_va.copy()
            pred['p_default'] = p_va
            pred['fold'],pred['sample'],pred['method'] = fold,key,method
            current.append(pred)
            if fold==1:
                art.update({'classifier':clf,'scaler':scaler,
                    'valid_rows':valid[['news_id','inn','text','lemma_text']].copy(),
                    'valid_news_vectors':valid_news,
                    'fit_docs':fit_docs if method=='w2v' else None})
                artifacts[(key,method)] = art
            print(f'{SAMPLE_LABELS[key]} | {METHOD_LABELS[method]} | фолд {fold}/{N_SPLITS} | '
                  f'Gini train={mt["gini"]:.3f}; valid={mv["gini"]:.3f} | {elapsed:.1f} c',flush=True)
            pd.DataFrame(fold_results).to_csv(RESULT_DIR/'fold_metrics.csv',index=False,encoding='utf-8-sig')
        oof = pd.concat(current,ignore_index=True).sort_values('inn')
        assert not oof.inn.duplicated().any() and len(oof)==df.inn.nunique()
        prediction_frames.append(oof)
        oof.to_csv(RESULT_DIR/f'oof_{key}_{method}.csv',index=False,encoding='utf-8-sig')
        mr = metrics(oof.target,oof.p_default)
        ff = pd.DataFrame([r for r in fold_results if r['sample']==key and r['method']==method])
        summaries.append({'sample':key,'method':method,'companies':len(oof),
            'default_rate':oof.target.mean(),'oof_gini':mr['gini'],
            'mean_fold_gini':ff.valid_gini.mean(),'std_fold_gini':ff.valid_gini.std(ddof=1),
            'oof_ap':mr['ap'],'mean_train_gini':ff.train_gini.mean()})
    view = pd.DataFrame([r for r in fold_results if r['method']==method])[
        ['sample','fold','train_gini','valid_gini','valid_ap','valid_no_text']].copy()
    view['sample'] = view['sample'].map(SAMPLE_LABELS)
    view.columns = ['Срез','Фолд','Gini train','Gini valid','AP valid','Без текстового сигнала']
    show_table(view,[25,9,16,16,15,19])

def show_topics(art,top_n=10):
    words = art['vectorizer'].get_feature_names_out()
    rows = [{'Тема':k,'Слова с наибольшими весами':', '.join(words[np.argsort(w)[::-1][:top_n]])}
            for k,w in enumerate(art['model'].components_)]
    show_table(pd.DataFrame(rows),[12,88])

def show_topic_documents(art,topic=0,top_n=3):
    rows = art['valid_rows'].copy()
    rows['topic_weight'] = art['valid_news_vectors'][:,topic]
    top = rows.loc[rows.topic_weight.gt(0)].drop_duplicates('news_id').nlargest(top_n,'topic_weight')
    show_table(pd.DataFrame({'ИНН':top.inn,'Вес темы':top.topic_weight,
        'Начало текста':top.text.map(lambda s:clean_text(s)[:350])}),[17,15,68])
````

## TF-IDF: веса слов в документе

### Мешок слов

Рассмотрим три новости:

1. «Компания погасила долг банку».
2. «Компания допустила просрочку долга».
3. «Компания открыла завод».

После лемматизации составим словарь всех встретившихся слов. Каждому слову соответствует столбец, каждой новости — строка. В ячейке можно записать число вхождений слова в документ. Такое представление называется **мешком слов**: оно сохраняет состав текста, но не полный порядок слов.

Слово «компания» присутствует во всех трёх сообщениях и почти не помогает их различать. Слово «просрочка» встречается только во втором. TF-IDF учитывает одновременно частоту слова внутри документа и его распространённость в корпусе.

### Частота и обратная документная частота

Обозначим число вхождений слова $t$ в документ $d$ через $c_{td}$. Для частоты используется логарифмическое преобразование:

$$tf_{td}=\begin{cases}1+\ln c_{td},&c_{td}>0,\\0,&c_{td}=0.\end{cases}$$

Одно вхождение даёт значение 1, десять — примерно 3.30. Повторение слова увеличивает его вес, но десятикратное повторение не делает слово в десять раз важнее.

Пусть $N$ — число обучающих документов, а $df_t$ — число документов, содержащих слово. Сглаженная обратная документная частота определяется как

$$idf_t=\ln\frac{1+N}{1+df_t}+1.$$

Здесь считается число документов со словом, а не суммарное число его повторений. В нашем примере для «компания» $N=3$ и $df=3$, поэтому $idf=1$. Для «просрочка» $df=1$, поэтому $idf=1+\ln2\approx1.693$.

Ненормированный вес равен произведению двух величин:

$$a_{td}=tf_{td}\,idf_t.$$

Затем вектор документа делится на его евклидову длину:

$$x_{td}=\frac{a_{td}}{\sqrt{\sum_u a_{ud}^2}}.$$

Нормировка уменьшает влияние различий в длине документов. Новость без известных слов сохраняет нулевой вектор. В числовом примере ниже показаны веса до нормировки, чтобы их можно было сопоставить с формулой.

### Словосочетания и близость документов

Кроме отдельных слов используются соседние пары лемм — биграммы. В тексте «не выплатить долг» появляются признаки «не», «выплатить», «долг», «не выплатить» и «выплатить долг». Биграмма помогает отличить отрицание от утверждения. Поскольку стоп-слова уже удалены, соседство определяется в обработанном тексте.

Близость двух векторов можно измерить косинусом:

$$\cos(x,y)=\frac{x^\top y}{\|x\|\,\|y\|}.$$

Для нормированных векторов это скалярное произведение. Документы с общими словами сближаются. Однако «предприятие погасило задолженность» и «компания вернула долг» могут остаться далеко друг от друга, хотя их смысл похож: TF-IDF самостоятельно не устанавливает синонимию.

В большой коллекции словарь может включать очень много признаков, но одна новость использует лишь небольшую их часть. Поэтому матрица хранится в разреженном виде — через ненулевые элементы. Число признаков определяется обучающим корпусом; редкие и частые слова после удаления стоп-слов сохраняются.

### Признаки дефолта

Векторы новостей усредняются по компаниям и поступают в логистическую регрессию. Словарь и IDF оцениваются только на обучающих текстах текущего фолда. Слова, которые впервые встретились в валидации, не добавляют новые столбцы.

Коэффициенты модели показывают, какие признаки связаны с более высоким или низким скором при прочих признаках. Положительный коэффициент слова «иск» не доказывает финансовое неблагополучие компании: она может взыскивать задолженность как истец. Помимо событий модель может использовать названия компаний, отраслевую лексику и особенности источников. Это одна из причин сравнивать качество на обучении и валидации.

## LDA: документ как смесь тем

### Что называется темой

В тематической модели тема — это распределение вероятностей по словам. В одной теме высокие вероятности могут получить «облигация», «купон», «эмитент», в другой — «суд», «иск», «взыскание», в третьей — «завод», «строительство», «оборудование».

Названия «Долговой рынок», «Судебные споры» и «Производство» присваивает исследователь после обучения. **В LDA не требуется заранее задавать перечень содержательных тем.** Задаётся их число $K$, а распределения слов оцениваются по корпусу. В эксперименте используются 20 тем.

Каждая новость может относиться к нескольким темам. Сообщение «Компания разместила облигации для строительства завода» сочетает долговую и производственную тематики. Условное представление $(0.65,0.05,0.30)$ означает, что документ описывается смесью трёх тем с такими весами. Это распределение тем, а не распределение вероятностей дефолта.

### Два набора распределений

Обозначим смесь тем документа $d$ через $\theta_d$, а распределение слов темы $k$ — через $\phi_k$. Тогда вероятность слова в документе выражается как

$$P(w\mid d)=\sum_{k=1}^{K}\theta_{dk}\phi_{kw}.$$

Предположим, вероятности слова «купон» в трёх темах равны $(0.08,0.001,0.002)$. При смеси документа $(0.65,0.05,0.30)$ получаем

$$P(\text{купон}\mid d)=0.65\cdot0.08+0.05\cdot0.001+0.30\cdot0.002=0.05265.$$

Таким образом, одно слово может встречаться в разных темах, но с разными вероятностями. LDA одновременно оценивает таблицу «документы × темы» и таблицу «темы × слова».

### Вероятностная модель появления текста

LDA предполагает следующий механизм: у документа есть собственная смесь тем; для каждой позиции выбирается тема из этой смеси, затем слово из распределения выбранной темы. Порядок слов в базовой модели не учитывается.

Распределение Дирихле задаёт априорные предпочтения для смесей:

$$\theta_d\sim Dirichlet(\alpha),\qquad \phi_k\sim Dirichlet(\eta).$$

При малых симметричных значениях $\alpha$ предпочтительнее смеси, сосредоточенные на немногих темах; при больших — более равномерные. Параметр $\eta$ аналогично влияет на распределение слов внутри темы. Эти предпочтения сочетаются с наблюдаемыми данными и не закрепляют за документом заранее определённые темы.

### Как находятся темы

Сами темы и назначения слов неизвестны. Обучение начинается с предварительных оценок. По текущим распределениям слов оценивается, какие темы могли объяснить каждый документ. Затем распределения слов тем уточняются: слова, которым приписана большая принадлежность к одной теме, получают в ней больший вес. Эти уточнения повторяются.

В реализации scikit-learn используется вариационный вывод — приближённая оценка скрытых распределений. При online-обучении общие параметры тем обновляются по последовательным группам документов. Группа определяет порядок вычислений, а не отбор части корпуса: в обучении участвуют все обучающие документы.

На вход LDA поступает матрица **числа вхождений слов**. Она соответствует предположению о генерации последовательности слов. TF-IDF решает другую задачу взвешивания и здесь не используется.

### От тем к дефолту

После обучения распределения слов тем фиксируются. Для новой новости оценивается только её смесь относительно найденных тем. Например, среднее трёх новостных векторов $(0.8,0.1,0.1)$, $(0.2,0.7,0.1)$ и $(0.5,0.2,0.3)$ равно $(0.5,0.333,0.167)$. Такой тематический профиль компании поступает в логистическую регрессию.

LDA не использует метки дефолта. Связь тем с дефолтом устанавливает классификатор. Тема судебных разбирательств может объединять публикации и об истцах, и об ответчиках; распределение тем не заменяет распознавание роли компании.

Если в новости не осталось известных слов, содержательного основания для оценки её темы нет. В таком случае используется нулевой вектор вместо априорной смеси. Номер темы также не имеет постоянного смысла: при повторном обучении тема 0 может описывать другое содержание. Поэтому профили тем разных фолдов нельзя сопоставлять только по номеру.

````python
#@title Таблица
show_table(pd.DataFrame([['K', 'Сколько скрытых распределений слов искать', 'Топ-слова и примеры документов каждой темы.'], ['α: doc_topic_prior', 'Предпочтение для смеси тем документа', 'Концентрацию документов на отдельных темах.'], ['η: topic_word_prior', 'Предпочтение для распределения слов темы', 'Концентрацию профиля темы на отдельных словах.'], ['max_iter', 'Число проходов обучения', 'Устойчивость тем при более длительном обучении.']], columns=['Что задаём', 'Что означает', 'Что смотрим после обучения']), widths=[25, 35, 40])
````

````python
#@title Модель 2: счётчики слов → LDA → темы компании → LR
def build_lda(fit_docs,train_docs,valid_docs):
    vectorizer = CountVectorizer(tokenizer=str.split,token_pattern=None,lowercase=False,
                                 min_df=1,max_df=1.0,max_features=None)
    X_fit = vectorizer.fit_transform(fit_docs)
    if min(X_fit.shape)<=N_TOPICS:
        raise ValueError('Мало документов или слов: уменьшите N_TOPICS.')
    lda = LatentDirichletAllocation(n_components=N_TOPICS,learning_method='online',
        batch_size=512,max_iter=LDA_ITER,max_doc_update_iter=LDA_DOC_ITER,evaluate_every=-1,
        doc_topic_prior=1/N_TOPICS,topic_word_prior=1/N_TOPICS,random_state=SEED,n_jobs=1)
    lda.fit(X_fit)
    def transform(documents):
        X = vectorizer.transform(documents)
        Z = lda.transform(X)
        Z[np.asarray(X.getnnz(axis=1)).ravel()==0] = 0
        return Z.astype(np.float32)
    return transform(train_docs),transform(valid_docs),{'vectorizer':vectorizer,'model':lda}
evaluate_method('lda',build_lda)
````

## NMF: разложение матрицы на неотрицательные компоненты

### Откуда берутся темы

Пусть $X$ — неотрицательная матрица TF-IDF: строки соответствуют документам, столбцы — словам. NMF ищет две меньшие матрицы, произведение которых приближает исходную:

$$X_{D\times V}\approx W_{D\times K}H_{K\times V},\qquad W\geq0,\quad H\geq0.$$

Строка $H$ описывает словесный профиль одного компонента. Строка $W$ показывает, с какими весами эти компоненты входят в документ. Компоненты можно интерпретировать как темы, если их характерные слова образуют содержательные группы.

Рассмотрим словарь «долг, купон, завод, оборудование». Пусть два профиля имеют вид

$$H_1=(0.5,0.5,0,0),\qquad H_2=(0,0,0.5,0.5).$$

Документ с весами $(2,0)$ восстанавливается как $(1,1,0,0)$, а документ с весами $(1,1)$ — как $(0.5,0.5,0.5,0.5)$. Первый связан с долгом, второй сочетает долговую и производственную тематики.

В этом примере профили заданы для иллюстрации умножения. При обучении алгоритм находит их самостоятельно. Если «долг» и «купон» часто встречаются вместе, общий компонент позволяет объяснить повторяющийся рисунок этих слов сразу во многих документах.

### Обучение NMF

Для отдельного элемента приближение имеет вид

$$\hat X_{dv}=\sum_{k=1}^{K}W_{dk}H_{kv}.$$

В рассматриваемом варианте минимизируется суммарная квадратичная ошибка восстановления:

$$\min_{W,H\geq0}\frac12\|X-WH\|_F^2
=\min_{W,H\geq0}\frac12\sum_{d,v}(X_{dv}-\hat X_{dv})^2.$$

Алгоритм по очереди уточняет веса документов и профили компонентов. При фиксированном $H$ ищутся более подходящие значения $W$, затем при фиксированном $W$ обновляется $H$. Неотрицательность позволяет складывать вклады компонентов, но запрещает компенсировать один компонент отрицательным весом другого.

Здесь используется координатный спуск: значения параметров последовательно изменяются так, чтобы уменьшать ошибку при сохранении неотрицательности. Начальное приближение строится на основе сингулярного разложения. Совместная задача по $W$ и $H$ невыпуклая, поэтому число компонентов и начальное приближение могут влиять на найденное решение.

### Веса и вероятности

У NMF допустимы веса $(2,1)$: они не обязаны суммироваться к единице. Более того, умножение строки $H_k$ на положительную константу и деление соответствующего столбца $W_k$ на неё не меняют произведение $WH$. Поэтому сравнение необработанных весов зависит от масштаба компонентов.

Чтобы получить относительные доли, сначала учтём сумму весов слов каждого профиля:

$$s_k=\sum_v H_{kv},\qquad \widetilde W_{dk}=W_{dk}s_k.$$

Это соответствует приведению каждого ненулевого профиля $H_k$ к сумме 1 с компенсацией масштаба в $W$. Затем нормируем вклады документа:

$$q_{dk}=\frac{\widetilde W_{dk}}{\sum_l\widetilde W_{dl}}.$$

Вектор $q_d$ удобно усреднять по новостям компании. Его элементы суммируются к единице, но остаются относительными весами компонентов разложения. Они не являются апостериорными вероятностями тем LDA.

### Прогноз по тематическим признакам

TF-IDF и NMF обучаются внутри каждого фолда. Для валидационных документов матрица $H$ уже фиксирована; оцениваются только веса сочетания найденных компонентов. После агрегации по компаниям признаки стандартизируются и передаются в логистическую регрессию.

LDA и NMF используют одинаковую предобработку и одинаковое число тем, но различаются исходными предположениями. LDA объясняет частоты слов вероятностной моделью, NMF приближает матрицу TF-IDF произведением неотрицательных матриц. Меньшая ошибка восстановления NMF ещё не означает более точного прогноза дефолта: восстановление текста и предсказание метки — разные задачи.

````python
#@title Модель 3: TF-IDF → NMF → относительные веса тем компании → LR
def build_nmf(fit_docs,train_docs,valid_docs):
    vectorizer = TfidfVectorizer(tokenizer=str.split,token_pattern=None,lowercase=False,
        min_df=1,max_df=1.0,max_features=None,sublinear_tf=True,norm='l2',dtype=np.float64)
    X_fit = vectorizer.fit_transform(fit_docs)
    if min(X_fit.shape)<=N_TOPICS:
        raise ValueError('Мало документов или слов: уменьшите N_TOPICS.')
    nmf = NMF(n_components=N_TOPICS,init='nndsvda',solver='cd',
              beta_loss='frobenius',max_iter=NMF_ITER,random_state=SEED)
    nmf.fit(X_fit)
    topic_mass = nmf.components_.sum(axis=1)
    def transform(documents):
        W = nmf.transform(vectorizer.transform(documents))
        return normalize(W*topic_mass[None,:],norm='l1').astype(np.float32)
    return transform(train_docs),transform(valid_docs),{
        'vectorizer':vectorizer,'model':nmf,'topic_mass':topic_mass}
evaluate_method('nmf',build_nmf)
````

## Эмбеддинги: Word2Vec и векторы новостей

### Почему близкие слова могут иметь близкие векторы

В TF-IDF слова «долг» и «задолженность» занимают независимые столбцы. **Эмбеддинг** представляет слово набором чисел, обученным так, чтобы отражать закономерности его употребления. Если слова встречаются в похожем окружении, их векторы могут оказаться близкими.

Word2Vec обучается на локальном контексте. Например, в последовательности «компания погасить долг банк» для центрального слова «долг» соседями будут «погасить» и «банк». Во многих новостях похожее окружение может встречаться и у слова «задолженность».

Близость по контексту не означает тождество смысла. «Прибыль» и «убыток» тоже могут находиться рядом в векторном пространстве, поскольку оба слова встречаются в сообщениях о финансовых результатах.

### Skip-gram и отрицательные примеры

В варианте **skip-gram** центральное слово используется для предсказания соседних слов. У слов есть обучаемые векторы; скалярное произведение векторов центрального слова и контекстного слова определяет оценку их совместимости.

Обрабатывать все слова словаря для каждой пары дорого. Negative sampling заменяет такую задачу различением наблюдаемой пары и нескольких случайно сформированных пар. Для пары $(w,c)$ максимизируется выражение вида

$$\log\sigma(u_c^\top v_w)+\sum_{r=1}^{m}\log\sigma(-u_{n_r}^\top v_w),$$

где $c$ — наблюдаемый сосед, $n_r$ — отрицательный пример, а $\sigma$ — логистическая функция. Пара «погасить — долг» должна получать более высокую совместимость, чем случайная пара. При обновлении векторов по множеству таких наблюдений формируется геометрия словаря.

Используются векторы длины 100 и контекстное окно до пяти слов с каждой стороны. Пять отрицательных примеров создаются для одной положительной пары. Обучение проходит десять эпох. Все встречающиеся обучающие слова включаются в словарь; частотное прореживание токенов отключено. Метки дефолта Word2Vec не использует.

### Вектор статьи

Word2Vec даёт векторы слов, а для классификации нужен вектор документа. Представим статью средним вектором её известных токенов:

$$z_d=\frac{1}{|T_d^{known}|}\sum_{w\in T_d^{known}}e(w).$$

Повторяющиеся слова учитываются столько раз, сколько встречаются в тексте. Если условно $e(\text{кредит})=(0.8,0.2)$ и $e(\text{долг})=(0.6,0.4)$, среднее для «кредит долг» равно $(0.7,0.3)$. После усреднения вектор нормируется по евклидовой длине. Новость без известных слов получает нулевой вектор.

Координаты эмбеддинга не имеют заранее заданных названий: первая координата не означает «негатив», а вторая — «позитив». Содержательно интерпретируется прежде всего взаимное положение векторов, например ближайшие соседи слова.

Векторы новостей усредняются по компании, затем стандартизируются и поступают в логистическую регрессию. Word2Vec обучается только на новостях обучающих компаний текущего фолда; при обработке валидации его словарь и векторы фиксированы.

### Что теряется при усреднении

У Word2Vec одно слово имеет один вектор независимо от конкретной статьи. Кроме того, усреднение не сохраняет порядок слов. Сообщения «Компания А подала иск к компании Б» и «Компания Б подала иск к компании А» могут получить одинаковое представление, хотя роли участников противоположны. В длинной статье краткое сообщение о просрочке также может ослабнуть среди многочисленных других слов.

Контекстные модели, включая трансформеры, учитывают окружение конкретного употребления слова. Это другой класс представлений: для готового текстового энкодера обычно используются исходный текст и его собственный токенизатор. В этом разделе изучается именно Word2Vec; его результаты не характеризуют все способы построения эмбеддингов.

### Кластеризация новостей

Плотные векторы можно использовать и без меток дефолта. KMeans разбивает новости на группы, минимизируя расстояния до центров кластеров:

$$\sum_d\|z_d-\mu_{c(d)}\|_2^2.$$

Алгоритм поочерёдно относит каждый вектор к ближайшему центру и пересчитывает центры как средние назначенных им векторов. В отличие от LDA, обычный KMeans присваивает новости один кластер, а не смесь тем.

Содержание кластера изучается по близким к центру публикациям. Номер кластера произволен и не обозначает уровень риска. Здесь кластеризация используется для изучения структуры текстов; классификатор дефолта работает с самими усреднёнными эмбеддингами. Центры кластеров оцениваются на обучающих текстах первого фолда, примеры берутся из его валидации.

````python
#@title Модель 4: Word2Vec → средний вектор новости → вектор компании → LR
def stable_hash(value):
    return int.from_bytes(hashlib.md5(str(value).encode()).digest()[:4],byteorder='little')

def mean_word_vectors(documents,keyed_vectors):
    result = np.zeros((len(documents),keyed_vectors.vector_size),dtype=np.float32)
    for i,text in enumerate(documents):
        ids = [keyed_vectors.key_to_index[w] for w in text.split() if w in keyed_vectors]
        if ids:
            result[i] = keyed_vectors.vectors[ids].mean(axis=0)
    return normalize(result,norm='l2')

class WordCorpus:
    def __init__(self, documents):
        self.documents = documents

    def __iter__(self):
        from gensim.models.word2vec import MAX_WORDS_IN_BATCH
        # Gensim обрезает слишком длинную последовательность внутри Cython.
        # Передаём её частями: все слова сохраняются, все части обучаются.
        for text in self.documents:
            tokens = text.split()
            for start in range(0, len(tokens), MAX_WORDS_IN_BATCH):
                yield tokens[start:start + MAX_WORDS_IN_BATCH]

def build_w2v(fit_docs,train_docs,valid_docs):
    model = Word2Vec(sentences=WordCorpus(fit_docs),vector_size=W2V_DIM,
        window=5,min_count=1,sample=0,sg=1,negative=5,epochs=W2V_EPOCHS,
        seed=SEED,workers=1,hashfxn=stable_hash)
    return (mean_word_vectors(train_docs.tolist(),model.wv),
            mean_word_vectors(valid_docs.tolist(),model.wv),{'model':model})
evaluate_method('w2v',build_w2v)
````
